```python
import pandas as pd
import numpy as np

df = pd.read_csv(r"C:\Users\HP\auspify-ml-internship\data\Dataset.csv")
print(df.shape)
df.head()
```

```python
df.info()
print(df.isna().sum())
print(df["type"].value_counts())
print(df["rating"].value_counts())
print("duplicate rows:", df.duplicated().sum())
print("duplicate ids:", df["show_id"].duplicated().sum())
```

```python
for col in ["director", "country", "date_added", "duration", "listed_in"]:
    print(col)
    print(df[col].value_counts().head(5))
    print()
```

```python
# 1. check whether "Not Given" hides elsewhere
print((df == "Not Given").sum())

# 2. replace it with "Unknown" in director and country
df["director"] = df["director"].replace("Not Given", "Unknown")
df["country"] = df["country"].replace("Not Given", "Unknown")

# 3. verify
print((df == "Not Given").sum().sum())
print(df["director"].value_counts().head(3))
```

```python
# 1. date_added -> real datetime, then extract year and month
df["date_added"] = pd.to_datetime(df["date_added"], format="%m/%d/%Y")
df["year_added"] = df["date_added"].dt.year
df["month_added"] = df["date_added"].dt.month

# 2. split duration into a number and a unit
df["duration_value"] = df["duration"].str.extract(r"(\d+)").astype(int)
df["duration_unit"] = df["duration"].str.extract(r"([A-Za-z]+)")

# 3. verify
print(df[["date_added", "year_added", "month_added", "duration", "duration_value", "duration_unit"]].head())
print(df["duration_unit"].value_counts())
print(df.groupby("type")["duration_unit"].value_counts())
```
